```python
#---------------------------------------ANSWER---------------------
##----------------------------------------------------------------------
#  Definition for a binary tree node. //Class TreeNode create binary tree
class TreeNode:
    def __init__(self, val=0, left=None, right=None): 
        self.val = val  # store actule number of the data
        self.left = left # left child
        self.right = right #right child
```

Note: 
 Depth-First Search (DFS) a tree by going deep into one branch first before visiting other branches and
 alla DFS go downwards first so it work Breadth-First Search (BFS)check levle by level

```python
from collections import deque

# Duplicate check using  Breadth-First Search (BFS) on the Tree
def is_duplicate(root: TreeNode) -> int:
    if root is None:
        return -1

    seen = set()            #  Set (Hash Table) for O(1) search stored visited value
    queue = deque([root])   #  Queus Breadth-First Search (BFS) , store nodes  and to visit next

    while queue:
        current_node = queue.popleft()  # run O(n) time ,one node at time

        if current_node.val in seen:    # Check if duplicate value exists in Hash Table
            return current_node.val     # closest duplicate check

        seen.add(current_node.val)     # store the value

        if current_node.left:           # add children to queue
            queue.append(current_node.left)

        if current_node.right:
            queue.append(current_node.right)     # Record  in Hash Table      
    return -1
```

```python
def list_to_tree(lst):
    if not lst:
        return None

    nodes = [TreeNode(val) if val is not None else None for val in lst]  # Create all nodes first
    it = iter(nodes)
    
    root = next(it)              # The first node is the root
   
    for node in nodes:           # Loop through the nodes and assign their children
        if node:
                                 # next(it, None) gets the next node, or None if the list is empty
            node.left = next(it, None)
            node.right = next(it, None)
            
    return root
```

```python
# --- Testing of the Example ---
# Example 1: root = [1, 2, 2, 3, 5, 6, 7] # Expected: 2
tree1 = list_to_tree([1, 2, 2, 3, 5, 6, 7])
print(f"Example 1 Output: {is_duplicate(tree1)}") 
```

```python
# Your answer here

def is_valid_brackets(s: str) -> bool:
    """
    Determine whether s contains a proper/valid bracket sequence.

    Time:  O(n)
    Space: O(n) worst-case (all openers)
    """
    valid_brackets = {'[': ']', '{': '}', '(': ')'}
    stack = []

    for c in s:
        # Opening bracket
        if c in valid_brackets:
            stack.append(c)
        # Closing bracket
        else:
            if not stack:
                return False
            last_open = stack.pop()
            if valid_brackets[last_open] != c:
                return False

    return len(stack) == 0


# Examples
tests = [
    ("{[()()[]]}", True),
    ("{[(])}", False),
    ("(((((((((())))))))))", True),
    ("((((((((", False)
]

for s, expected in tests:
    result = is_valid_brackets(s)
    print(f"Input: {s}  Output: {result}  Expected: {expected}")
```

```python
# Your answer here
import timeit
time_complexity = timeit.timeit(lambda: is_valid_brackets('{}'), number=1000) 
time_complexity
```

Time Complexity: O(n) Each character is processed once, and each bracket is pushed and popped at most one time.

Space Complexity: O(n) In the worst case like "((((((", all characters are opening brackets and must be stored in the stack.
